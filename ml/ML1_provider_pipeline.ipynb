```python
# ============================================================
# GEOSPATIAL PREPARATION - INSPECTION
# ============================================================

import pandas as pd
import numpy as np

CLEAN_FILE = (
    "/content/drive/MyDrive/"
    "Provider_Network_Project/"
    "data/processed/"
    "provider_clean_300000.csv"
)

df = pd.read_csv(
    CLEAN_FILE,
    dtype=str,
    low_memory=False
)

print("=" * 70)
print("GEOSPATIAL PREPARATION")
print("=" * 70)

print("\nDataset shape:")
print(df.shape)

# ------------------------------------------------------------
# Address columns
# ------------------------------------------------------------

address_columns = [
    "Provider First Line Business Practice Location Address",
    "Provider Second Line Business Practice Location Address",
    "Provider Business Practice Location Address City Name",
    "Provider Business Practice Location Address State Name",
    "Provider Business Practice Location Address Postal Code"
]

print("\nAddress missingness:")

for col in address_columns:
    missing = df[col].isna().sum()
    percentage = missing / len(df) * 100

    print(
        f"{col}: "
        f"{missing:,} missing "
        f"({percentage:.2f}%)"
    )

# ------------------------------------------------------------
# Geospatial eligibility
# ------------------------------------------------------------

print("\nGEOSPATIAL_READY distribution:")
print(
    df["GEOSPATIAL_READY"]
    .value_counts(dropna=False)
)

print("\nADDRESS_READY distribution:")
print(
    df["ADDRESS_READY"]
    .value_counts(dropna=False)
)

# ------------------------------------------------------------
# Unique ZIPs
# ------------------------------------------------------------

print("\nUnique ZIP codes:")

unique_zips = (
    df.loc[
        df["ZIP_CLEAN"].notna(),
        "ZIP_CLEAN"
    ]
    .nunique()
)

print(unique_zips)

# ------------------------------------------------------------
# Unique state + ZIP combinations
# ------------------------------------------------------------

unique_state_zip = (
    df.loc[
        df["GEOSPATIAL_READY"] == True,
        ["STATE_CLEAN", "ZIP_CLEAN"]
    ]
    .drop_duplicates()
    .shape[0]
)

print(
    "\nUnique State + ZIP combinations:",
    unique_state_zip
)

# ------------------------------------------------------------
# Exact address uniqueness
# ------------------------------------------------------------

geo_df = df[
    df["GEOSPATIAL_READY"] == True
].copy()

geo_df["ADDRESS_KEY"] = (
    geo_df[
        "Provider First Line Business Practice Location Address"
    ]
    .fillna("")
    .str.upper()
    .str.strip()
    + "|"
    +
    geo_df[
        "Provider Business Practice Location Address City Name"
    ]
    .fillna("")
    .str.upper()
    .str.strip()
    + "|"
    +
    geo_df["STATE_CLEAN"]
    + "|"
    +
    geo_df["ZIP_CLEAN"]
)

unique_addresses = (
    geo_df["ADDRESS_KEY"]
    .nunique()
)

print(
    "\nProviders with usable geography:",
    f"{len(geo_df):,}"
)

print(
    "Unique practice addresses:",
    f"{unique_addresses:,}"
)

print(
    "\nPotential geocoding reduction:",
    f"{len(geo_df) - unique_addresses:,}",
    "duplicate address occurrences"
)
```

```python
# ============================================================
# FIX GEOSPATIAL INSPECTION
# ============================================================

# Convert GEOSPATIAL_READY from string to real boolean
df["GEOSPATIAL_READY_BOOL"] = (
    df["GEOSPATIAL_READY"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("true")
)

df["ADDRESS_READY_BOOL"] = (
    df["ADDRESS_READY"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("true")
)

# ------------------------------------------------------------
# Providers with usable geography
# ------------------------------------------------------------

geo_df = df[
    df["GEOSPATIAL_READY_BOOL"]
].copy()

print("=" * 70)
print("CORRECTED GEOSPATIAL INSPECTION")
print("=" * 70)

print(
    "\nProviders with usable geography:",
    f"{len(geo_df):,}"
)

# ------------------------------------------------------------
# Unique ZIP codes
# ------------------------------------------------------------

unique_zips = (
    geo_df["ZIP_CLEAN"]
    .dropna()
    .nunique()
)

print(
    "Unique ZIP codes:",
    f"{unique_zips:,}"
)

# ------------------------------------------------------------
# Unique STATE + ZIP
# ------------------------------------------------------------

unique_state_zip = (
    geo_df[
        ["STATE_CLEAN", "ZIP_CLEAN"]
    ]
    .dropna()
    .drop_duplicates()
    .shape[0]
)

print(
    "Unique State + ZIP combinations:",
    f"{unique_state_zip:,}"
)

# ------------------------------------------------------------
# Create address key
# ------------------------------------------------------------

geo_df["ADDRESS_KEY"] = (
    geo_df[
        "Provider First Line Business Practice Location Address"
    ]
    .fillna("")
    .astype(str)
    .str.upper()
    .str.strip()
    + "|"
    +
    geo_df[
        "Provider Business Practice Location Address City Name"
    ]
    .fillna("")
    .astype(str)
    .str.upper()
    .str.strip()
    + "|"
    +
    geo_df["STATE_CLEAN"]
    .fillna("")
    .astype(str)
    .str.upper()
    .str.strip()
    + "|"
    +
    geo_df["ZIP_CLEAN"]
    .fillna("")
    .astype(str)
    .str.strip()
)

unique_addresses = (
    geo_df["ADDRESS_KEY"]
    .nunique()
)

print(
    "Unique practice addresses:",
    f"{unique_addresses:,}"
)

# ------------------------------------------------------------
# Duplicate address occurrences
# ------------------------------------------------------------

duplicate_address_occurrences = (
    len(geo_df) - unique_addresses
)

print(
    "Duplicate address occurrences:",
    f"{duplicate_address_occurrences:,}"
)

# ------------------------------------------------------------
# State distribution
# ------------------------------------------------------------

print("\nTop states:")

print(
    geo_df["STATE_CLEAN"]
    .value_counts()
    .head(20)
)

# ------------------------------------------------------------
# ZIP distribution
# ------------------------------------------------------------

print("\nTop ZIP codes:")

print(
    geo_df["ZIP_CLEAN"]
    .value_counts()
    .head(20)
)
```

```python
# ============================================================
# ML-1 GEOSPATIAL FEATURE ENGINEERING
# ZIP / STATE LEVEL
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("BUILDING GEOSPATIAL PROVIDER FEATURES")
print("=" * 70)

# ------------------------------------------------------------
# Make sure geography flags are boolean
# ------------------------------------------------------------

df["GEOSPATIAL_READY_BOOL"] = (
    df["GEOSPATIAL_READY"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("true")
)

# ------------------------------------------------------------
# Keep providers with usable geography
# ------------------------------------------------------------

geo_df = df[
    df["GEOSPATIAL_READY_BOOL"]
].copy()

print(
    "Providers entering geographic feature engineering:",
    f"{len(geo_df):,}"
)

# ------------------------------------------------------------
# Clean state / ZIP
# ------------------------------------------------------------

geo_df["STATE_CLEAN"] = (
    geo_df["STATE_CLEAN"]
    .astype(str)
    .str.strip()
    .str.upper()
)

geo_df["ZIP_CLEAN"] = (
    geo_df["ZIP_CLEAN"]
    .astype(str)
    .str.strip()
)

# ------------------------------------------------------------
# Create geographic key
# ------------------------------------------------------------

geo_df["GEO_KEY"] = (
    geo_df["STATE_CLEAN"]
    + "_"
    + geo_df["ZIP_CLEAN"]
)

# ------------------------------------------------------------
# Provider count
# ------------------------------------------------------------

provider_count = (
    geo_df
    .groupby(
        ["STATE_CLEAN", "ZIP_CLEAN"],
        as_index=False
    )
    .agg(
        provider_count=("NPI", "nunique")
    )
)

print(
    "\nGeographic areas created:",
    f"{len(provider_count):,}"
)

print("\nSample:")
print(provider_count.head(10))
```

```python
# ============================================================
# PROVIDER TYPE FEATURES
# ============================================================

provider_type_features = (
    geo_df
    .pivot_table(
        index=["STATE_CLEAN", "ZIP_CLEAN"],
        columns="PROVIDER_TYPE",
        values="NPI",
        aggfunc="nunique",
        fill_value=0
    )
    .reset_index()
)

# Rename columns safely
provider_type_features = provider_type_features.rename(
    columns={
        "Individual": "individual_provider_count",
        "Organization": "organization_provider_count",
        "Unknown": "unknown_provider_count"
    }
)

print("=" * 70)
print("PROVIDER TYPE FEATURES")
print("=" * 70)

print(provider_type_features.head())
```

```python
# ============================================================
# SPECIALTY FEATURES
# ============================================================

specialty_features = (
    geo_df
    .groupby(
        ["STATE_CLEAN", "ZIP_CLEAN"],
        as_index=False
    )
    .agg(
        unique_specialty_count=(
            "PRIMARY_TAXONOMY",
            lambda x: x.dropna().nunique()
        ),
        providers_with_specialty=(
            "PRIMARY_TAXONOMY",
            lambda x: x.notna().sum()
        )
    )
)

print("=" * 70)
print("SPECIALTY FEATURES")
print("=" * 70)

print(specialty_features.head(10))
```

```python
print(geo_df.columns.tolist())
```

```python
# ============================================================
# ZIP + SPECIALTY LEVEL FEATURES
# ============================================================

area_specialty_features = (
    geo_df
    .dropna(subset=["PRIMARY_TAXONOMY"])
    .groupby(
        ["STATE_CLEAN", "ZIP_CLEAN", "PRIMARY_TAXONOMY"],
        as_index=False
    )
    .agg(
        specialty_provider_count=(
            "NPI",
            "nunique"
        )
    )
)

print("=" * 70)
print("ZIP + SPECIALTY LEVEL FEATURES")
print("=" * 70)

print(area_specialty_features.head(20))
print("\nShape:", area_specialty_features.shape)
```

```python
# ============================================================
# MERGE GEOGRAPHIC FEATURES
# ============================================================

geo_features = provider_count.merge(
    provider_type_features,
    on=["STATE_CLEAN", "ZIP_CLEAN"],
    how="left"
)

geo_features = geo_features.merge(
    specialty_features,
    on=["STATE_CLEAN", "ZIP_CLEAN"],
    how="left"
)

print("=" * 70)
print("MERGED GEOSPATIAL FEATURES")
print("=" * 70)

print("Shape:", geo_features.shape)

print("\nColumns:")
print(geo_features.columns.tolist())

print("\nPreview:")
display(geo_features.head(10))
```

```python
# ============================================================
# SPECIALTY DIVERSITY
# ============================================================

geo_features["specialty_diversity_ratio"] = (
    geo_features["unique_specialty_count"]
    /
    geo_features["provider_count"]
)

geo_features["specialty_diversity_ratio"] = (
    geo_features["specialty_diversity_ratio"]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

print(
    geo_features[
        [
            "STATE_CLEAN",
            "ZIP_CLEAN",
            "provider_count",
            "unique_specialty_count",
            "specialty_diversity_ratio"
        ]
    ].head(10)
)
```

```python
# ============================================================
# STATE-LEVEL CONTEXT
# ============================================================

state_provider_totals = (
    geo_df
    .groupby("STATE_CLEAN")["NPI"]
    .nunique()
    .reset_index()
)

state_provider_totals = state_provider_totals.rename(
    columns={
        "NPI": "state_provider_count"
    }
)

geo_features = geo_features.merge(
    state_provider_totals,
    on="STATE_CLEAN",
    how="left"
)

geo_features["zip_provider_share_of_state"] = (
    geo_features["provider_count"]
    /
    geo_features["state_provider_count"]
)

print(
    geo_features[
        [
            "STATE_CLEAN",
            "ZIP_CLEAN",
            "provider_count",
            "state_provider_count",
            "zip_provider_share_of_state"
        ]
    ].head(10)
)
```

```python
# ============================================================
# PROVIDER COMPOSITION
# ============================================================

geo_features["individual_provider_ratio"] = (
    geo_features["individual_provider_count"]
    /
    geo_features["provider_count"]
)

geo_features["organization_provider_ratio"] = (
    geo_features["organization_provider_count"]
    /
    geo_features["provider_count"]
)

geo_features[
    [
        "individual_provider_ratio",
        "organization_provider_ratio"
    ]
] = (
    geo_features[
        [
            "individual_provider_ratio",
            "organization_provider_ratio"
        ]
    ]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

print(
    geo_features[
        [
            "STATE_CLEAN",
            "ZIP_CLEAN",
            "provider_count",
            "individual_provider_ratio",
            "organization_provider_ratio"
        ]
    ].head(10)
)
```

```python
# ============================================================
# STEP 9C — EXTRACT ZCTA / ZIP POPULATION
# ============================================================

import pandas as pd
from io import StringIO

print("=" * 70)
print("STEP 9C — EXTRACTING ZCTA POPULATION")
print("=" * 70)

# ------------------------------------------------------------
# Read Census pipe-delimited data
# ------------------------------------------------------------

population_raw = pd.read_csv(
    StringIO(text),
    sep="|",
    dtype=str
)

print("Raw rows:", len(population_raw))
print("Raw columns:", population_raw.columns.tolist())

# ------------------------------------------------------------
# Keep required columns
# ------------------------------------------------------------

population_df = population_raw[
    ["GEO_ID", "B01003_E001"]
].copy()

# ------------------------------------------------------------
# Identify ZCTA records
# ------------------------------------------------------------
# ZCTA GEO_ID values are represented as:
#
# 8600000USXXXXX
#
# where XXXXX is the 5-digit ZCTA.
#
# ------------------------------------------------------------

zcta_mask = (
    population_df["GEO_ID"]
    .str.startswith("8600000US", na=False)
)

population_df = population_df[zcta_mask].copy()

print("\nZCTA rows:", len(population_df))

# ------------------------------------------------------------
# Extract 5-digit ZIP/ZCTA
# ------------------------------------------------------------

population_df["ZIP_CLEAN"] = (
    population_df["GEO_ID"]
    .str.replace("8600000US", "", regex=False)
    .str.zfill(5)
)

# ------------------------------------------------------------
# Rename population column
# ------------------------------------------------------------

population_df = population_df.rename(
    columns={
        "B01003_E001": "population"
    }
)

# ------------------------------------------------------------
# Convert population to numeric
# ------------------------------------------------------------

population_df["population"] = pd.to_numeric(
    population_df["population"],
    errors="coerce"
)

# ------------------------------------------------------------
# Keep final columns
# ------------------------------------------------------------

population_df = population_df[
    ["ZIP_CLEAN", "population"]
].copy()

# ------------------------------------------------------------
# Remove invalid population values
# ------------------------------------------------------------

population_df = population_df[
    population_df["population"].notna()
].copy()

# ------------------------------------------------------------
# Remove duplicate ZIPs
# ------------------------------------------------------------

population_df = population_df.drop_duplicates(
    subset=["ZIP_CLEAN"]
)

# ------------------------------------------------------------
# Reset index
# ------------------------------------------------------------

population_df = population_df.reset_index(drop=True)

# ------------------------------------------------------------
# Inspect
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CENSUS ZCTA POPULATION DATA")
print("=" * 70)

print("Rows:", len(population_df))
print("Columns:", population_df.columns.tolist())

print(
    "Population missing:",
    population_df["population"].isna().sum()
)

print(
    "ZIP missing:",
    population_df["ZIP_CLEAN"].isna().sum()
)

print("\nSample:")
display(population_df.head(10))
```

```python
%cd /content/provider-network-adequacy-intelligence

documentation = """# ML-1 Data Engineering and Geospatial Feature Engineering

## Project
Provider Network Adequacy and Access Intelligence

## Role
ML-1: Data Engineering and Geospatial Feature Engineering

## Objective
Prepare a clean, aggregated, geographic and ML-ready dataset for downstream machine learning.

## Data Sources

### 1. NPPES NPI Registry
Used for provider information, provider type, specialty, state and ZIP code.

The full NPPES file is very large, so a 300,000-record working sample was used for the hackathon pipeline.

### 2. Census ACS 2023 5-Year
Used to obtain population estimates at the ZCTA/ZIP level.

Dataset:
B01003 - Total Population

## Pipeline

NPPES Provider Data
        ↓
Sampling
        ↓
Data Inspection
        ↓
Data Cleaning
        ↓
Provider Type Processing
        ↓
Specialty Processing
        ↓
ZIP/State Standardization
        ↓
ZIP-Level Provider Aggregation
        ↓
Census Population Integration
        ↓
Provider Density Features
        ↓
EDA
        ↓
ML-Ready Dataset

## Sampling

A 300,000-record sample of NPPES was used because the complete NPPES monthly file is several GB in size.

The sampling approach was selected to make the pipeline practical to execute in Google Colab.

Important limitation:
The resulting dataset should be considered a hackathon working dataset and is not guaranteed to represent the complete US provider network.

## Geographic Feature Engineering

Providers were aggregated at ZIP level.

Generated geographic/provider features include:

- Provider count
- Individual provider count
- Organization provider count
- Unique specialty count
- Specialty diversity ratio
- State provider count
- ZIP share of state providers
- Individual provider ratio
- Organization provider ratio

## Population Integration

2023 Census ACS 5-Year ZCTA population data was joined using standardized ZIP/ZCTA identifiers.

Provider ZIP records: 16,934

Population matched: 16,202

Population match rate: 95.68%

Population records without a Census match: 732

Missing population values were retained as missing values and were NOT replaced with zero.

## Density Features

The following features were created:

- Providers per 10,000 population
- Individual providers per 10,000 population
- Organization providers per 10,000 population

Density features are calculated only where population is available and greater than zero.

## Final ML-Ready Dataset

File:

data/processed/provider_network_ml_ready.csv

Rows: 16,934

Features: 16

Duplicate rows: 0

## Final Features

1. STATE_CLEAN
2. ZIP_CLEAN
3. provider_count
4. individual_provider_count
5. organization_provider_count
6. unique_specialty_count
7. providers_with_specialty
8. specialty_diversity_ratio
9. state_provider_count
10. zip_provider_share_of_state
11. individual_provider_ratio
12. organization_provider_ratio
13. population
14. providers_per_10000
15. individual_providers_per_10000
16. organization_providers_per_10000

## EDA

Exploratory Data Analysis was performed on:

- Provider count distribution
- Provider density distribution
- Specialty diversity
- Population distribution
- Geographic/provider feature distributions

The distributions show strong skew in several provider and population-related variables.

## Important Limitations

1. NPPES data was sampled rather than using the complete provider registry.
2. ZIP code and Census ZCTA are related but are not exactly identical geographic concepts.
3. Some ZIP records do not have a matching Census population value.
4. Provider density can contain extreme values in areas with very small populations.
5. Provider count does not directly measure appointment availability, provider capacity, wait time or quality.
6. Target definition and final ML model selection are outside the ML-1 responsibility.

## Handoff to ML-2

The final ML-ready dataset is prepared for downstream:

- Target definition
- Feature selection
- Model development
- Model evaluation
- Provider access-gap prediction

ML-2 should consider appropriate handling of skewed density variables, including log transformation, robust scaling or winsorization where appropriate.

## Files

- ml/ML1_provider_pipeline.ipynb
- data/processed/provider_network_ml_ready.csv
- data/processed/provider_geo_features_zip.csv
- data/processed/census_zip_population_2023.csv
- data/processed/ML1_data_dictionary.csv
- data/processed/ML1_summary.txt

## Raw Data Policy

Large raw NPPES files and large intermediate files are intentionally not committed to GitHub.

They remain in Google Drive/local storage so that the GitHub repository remains lightweight.

"""

with open("docs/ML1_DATA_PIPELINE.md", "w") as f:
    f.write(documentation)

print("✅ ML-1 documentation created:")
print("docs/ML1_DATA_PIPELINE.md")

print("\n===== DOCUMENTATION =====")
!ls -lh docs/
```
